```sql
CREATE CATALOG IF NOT EXISTS banking;
CREATE SCHEMA IF NOT EXISTS banking.bronze;
```

```sql
CREATE TABLE IF NOT EXISTS banking.bronze.accounts (
    account_id BIGINT,
    customer_id BIGINT,
    branch_id BIGINT,
    account_type STRING,
    balance DECIMAL(18,2),
    open_date DATE,
    status STRING,
    updated_at TIMESTAMP,
    _rescued_data STRING
)
USING DELTA
LOCATION "abfss://banking-transaction-monitoring@project1storageacc.dfs.core.windows.net/bronze/accounts";
```

```sql
CREATE TABLE IF NOT EXISTS banking.bronze.branches (
    branch_id BIGINT,
    branch_name STRING,
    city STRING,
    state STRING,
    opened_date DATE,
    ifsc_code STRING,
    updated_at TIMESTAMP,
    _rescued_data STRING
)
USING DELTA
LOCATION "abfss://banking-transaction-monitoring@project1storageacc.dfs.core.windows.net/bronze/branches";
```

```sql
CREATE TABLE IF NOT EXISTS banking.bronze.card_transactions (
    card_txn_id BIGINT,
    card_id BIGINT,
    txn_date DATE,
    merchant_category STRING,
    amount DECIMAL(18,2),
    is_fraud BOOLEAN,
    updated_at TIMESTAMP,
    _rescued_data STRING
)
USING DELTA
LOCATION "abfss://banking-transaction-monitoring@project1storageacc.dfs.core.windows.net/bronze/card_transactions";
```

```sql
CREATE TABLE IF NOT EXISTS banking.bronze.cards (
    card_id BIGINT,
    customer_id BIGINT,
    account_id BIGINT,
    card_type STRING,
    issue_date DATE,
    expiry_date DATE,
    credit_limit DECIMAL(18,2),
    status STRING,
    updated_at TIMESTAMP,
    _rescued_data STRING
)
USING DELTA
LOCATION "abfss://banking-transaction-monitoring@project1storageacc.dfs.core.windows.net/bronze/cards";
```

```sql
CREATE TABLE IF NOT EXISTS banking.bronze.customers (
    customer_id BIGINT,
    name STRING,
    gender STRING,
    date_of_birth DATE,
    city STRING,
    state STRING,
    phone STRING,
    email STRING,
    occupation STRING,
    annual_income DECIMAL(18,2),
    join_date DATE,
    credit_score INT,
    updated_at TIMESTAMP,
    _rescued_data STRING
)
USING DELTA
LOCATION "abfss://banking-transaction-monitoring@project1storageacc.dfs.core.windows.net/bronze/customers";
```

```sql
CREATE TABLE IF NOT EXISTS banking.bronze.employees (
    employee_id BIGINT,
    name STRING,
    branch_id BIGINT,
    role STRING,
    hire_date DATE,
    salary DECIMAL(18,2),
    updated_at TIMESTAMP,
    _rescued_data STRING
)
USING DELTA
LOCATION "abfss://banking-transaction-monitoring@project1storageacc.dfs.core.windows.net/bronze/employees";
```

```sql
CREATE TABLE IF NOT EXISTS banking.bronze.loan_payments (
    payment_id BIGINT,
    loan_id BIGINT,
    payment_date DATE,
    amount_paid DECIMAL(18,2),
    principal_component DECIMAL(18,2),
    interest_component DECIMAL(18,2),
    late_payment_flag BOOLEAN,
    updated_at TIMESTAMP,
    _rescued_data STRING
)
USING DELTA
LOCATION "abfss://banking-transaction-monitoring@project1storageacc.dfs.core.windows.net/bronze/loan_payments";
```

```sql
CREATE TABLE IF NOT EXISTS banking.bronze.loans (
    loan_id BIGINT,
    customer_id BIGINT,
    branch_id BIGINT,
    loan_type STRING,
    loan_amount DECIMAL(18,2),
    interest_rate DECIMAL(5,2),
    term_months INT,
    start_date DATE,
    status STRING,
    updated_at TIMESTAMP,
    _rescued_data STRING
)
USING DELTA
LOCATION "abfss://banking-transaction-monitoring@project1storageacc.dfs.core.windows.net/bronze/loans";
```

```sql
CREATE TABLE IF NOT EXISTS banking.bronze.support_tickets (
    ticket_id BIGINT,
    customer_id BIGINT,
    issue_type STRING,
    date_opened DATE,
    date_resolved DATE,
    status STRING,
    satisfaction_score INT,
    updated_at TIMESTAMP,
    _rescued_data STRING
)
USING DELTA
LOCATION "abfss://banking-transaction-monitoring@project1storageacc.dfs.core.windows.net/bronze/support_tickets";
```

```sql
CREATE TABLE IF NOT EXISTS banking.bronze.transactions (
    transaction_id BIGINT,
    account_id BIGINT,
    txn_date DATE,
    txn_type STRING,
    amount DECIMAL(18,2),
    channel STRING,
    merchant_category STRING,
    updated_at TIMESTAMP,
    _rescued_data STRING
)
USING DELTA
LOCATION "abfss://banking-transaction-monitoring@project1storageacc.dfs.core.windows.net/bronze/transactions";
```
